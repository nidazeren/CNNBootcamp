```python
# Gerekli kütüphaneleri içe aktarma
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)
from tensorflow.keras.optimizers import Adam, RMSprop, SGD
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
import os
import matplotlib.cm as cm

# Hiperparametreler için varsayılan değerler
IMG_SIZE = (180, 180)
BATCH_SIZE = 32
EPOCHS = 50

print("✅ Kütüphaneler ve sabit hiperparametreler yüklendi.")

# --- 1. Veri Önişleme ---
DATASET_PATH = "/kaggle/input/brain-tumor-mri-dataset/"

training_path = os.path.join(DATASET_PATH, 'Training')
testing_path = os.path.join(DATASET_PATH, 'Testing')

if not os.path.exists(training_path):
    print(f"❌ Hata: '{training_path}' klasörü bulunamadı. Lütfen dosya yolunuzu ve klasör adını kontrol edin.")
    exit()
if not os.path.exists(testing_path):
    print(f"❌ Hata: '{testing_path}' klasörü bulunamadı. Lütfen dosya yolunuzu ve klasör adını kontrol edin.")
    exit()

try:
    ds_train = tf.keras.utils.image_dataset_from_directory(
        training_path,
        labels='inferred',
        label_mode='int',
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        validation_split=0.2,
        subset='training',
        seed=123
    )

    ds_val = tf.keras.utils.image_dataset_from_directory(
        training_path,
        labels='inferred',
        label_mode='int',
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        validation_split=0.2,
        subset='validation',
        seed=123
    )

    class_names = ds_train.class_names
    num_classes = len(class_names)
    print(f"✅ Veri seti yüklendi. Sınıf Sayısı: {num_classes}")

except Exception as e:
    print(f"Hata: Veri seti yüklenirken beklenmedik bir sorun oluştu.")
    print(f"Hata Detayı: {e}")
    exit()

# Veri Ön işleme ve Augmentation Fonksiyonları
def preprocess_train(image, label):
    image = tf.image.random_flip_left_right(image)
    image = tf.image.random_brightness(image, max_delta=0.1)
    image = tf.image.random_contrast(image, lower=0.9, upper=1.2)
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

def preprocess_val(image, label):
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

ds_train = (
    ds_train
    .map(preprocess_train, num_parallel_calls=tf.data.AUTOTUNE)
    .cache()
    .prefetch(buffer_size=tf.data.AUTOTUNE)
)
ds_val = (
    ds_val
    .map(preprocess_val, num_parallel_calls=tf.data.AUTOTUNE)
    .cache()
    .prefetch(buffer_size=tf.data.AUTOTUNE)
)
print("✅ Veri ön işleme ve augmentasyon tamamlandı.")

# --- 2. Modeli Oluşturan ve Eğiten Fonksiyon ---
def create_and_train_model(
    num_layers=3,
    filters=(32, 64, 128),
    kernel_size=(3, 3),
    dropout_rate=0.5,
    dense_units=128,
    learning_rate=0.001,
    optimizer_name='Adam'
):
    """
    Belirtilen hiperparametrelerle bir CNN modeli oluşturur ve eğitir.
    
    Args:
        num_layers (int): Evrişim katmanlarının sayısı.
        filters (tuple): Her katman için kullanılacak filtre sayısı.
        kernel_size (tuple): Her evrişim katmanı için kernel boyutu.
        dropout_rate (float): Dropout katmanının oranı.
        dense_units (int): Gizli Dense katmanındaki nöron sayısı.
        learning_rate (float): Öğrenme oranı.
        optimizer_name (str): Kullanılacak optimizer adı ('Adam', 'RMSprop', 'SGD').
        
    Returns:
        tuple: (Eğitilmiş model, eğitim geçmişi)
    """
    # Optimizer seçimi
    if optimizer_name == 'Adam':
        optimizer = Adam(learning_rate=learning_rate)
    elif optimizer_name == 'RMSprop':
        optimizer = RMSprop(learning_rate=learning_rate)
    elif optimizer_name == 'SGD':
        optimizer = SGD(learning_rate=learning_rate)
    else:
        raise ValueError("Geçersiz optimizer adı. 'Adam', 'RMSprop' veya 'SGD' kullanın.")

    # Modelin oluşturulması
    model = Sequential()
    
    # Evrişim ve Havuzlama Katmanları
    for i in range(num_layers):
        model.add(Conv2D(
            filters=filters[i],
            kernel_size=kernel_size,
            activation='relu',
            input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3) if i == 0 else None,
            name=f'conv_{i+1}'
        ))
        # Düzeltme: MaxPooling2D parantez hatası giderildi.
        model.add(MaxPooling2D((2, 2)))
    
    # Tam Bağlantılı Katmanlar
    model.add(Flatten())
    model.add(Dense(dense_units, activation='relu'))
    model.add(Dropout(dropout_rate))
    model.add(Dense(num_classes, activation='softmax'))
    
    # Modeli derleme
    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )
    print("✅ Model oluşturuldu ve derlendi.")
    model.build(input_shape=(None, IMG_SIZE[0], IMG_SIZE[1], 3))
    model.summary()

    # Modelin eğitilmesi
    callbacks = [
        EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True),
        ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=0.00001),
        ModelCheckpoint(filepath='best_model.keras', save_best_only=True)
    ]

    print("🚀 Model eğitimi başlıyor...")
    history = model.fit(
        ds_train,
        validation_data=ds_val,
        epochs=EPOCHS,
        callbacks=callbacks
    )
    print("✅ Model eğitimi tamamlandı.")
    
    return model, history

# --- 3. Deneme Yapılacak Parametreler ---
print("\n--- Deneme 1 Başlıyor: Varsayılan Parametreler ---")
current_model, history = create_and_train_model()


# --- 4. Modelin Değerlendirilmesi ---
print("\n✅ Model değerlendirmesi başlıyor...")

try:
    ds_test = tf.keras.utils.image_dataset_from_directory(
        testing_path,
        labels='inferred',
        label_mode='int',
        image_size=IMG_SIZE,
        batch_size=BATCH_SIZE,
        shuffle=False
    )
    ds_test = ds_test.map(lambda x, y: (tf.cast(x, tf.float32) / 255.0, y))

    loss, accuracy = current_model.evaluate(ds_test)
    print(f"Test Kaybı (Loss): {loss:.4f}")
    print(f"Test Başarım (Accuracy): {accuracy:.4f}")

    val_labels = np.concatenate([y for x, y in ds_test], axis=0)
    predictions = np.argmax(current_model.predict(ds_test), axis=1)

    print("\n--- Sınıflandırma Raporu ---")
    print(classification_report(val_labels, predictions, target_names=class_names))

    print("\n--- Karmaşıklık Matrisi ---")
    cm = confusion_matrix(val_labels, predictions)
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
    plt.xlabel('Tahmin Edilen Sınıf')
    plt.ylabel('Gerçek Sınıf')
    plt.title('Karmaşıklık Matrisi')
    plt.show()

except Exception as e:
    print(f"Hata: Test seti yüklenirken veya değerlendirilirken bir sorun oluştu.")
    print(f"Hata Detayı: {e}")

# Overfitting/Underfitting grafikleri
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(len(acc))

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Eğitim Başarımı')
plt.plot(epochs_range, val_acc, label='Doğrulama Başarımı')
plt.legend(loc='lower right')
plt.title('Eğitim ve Doğrulama Başarımı')
plt.xlabel('Epok')
plt.ylabel('Başarım (Accuracy)')
plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Eğitim Kaybı')
plt.plot(epochs_range, val_loss, label='Doğrulama Kaybı')
plt.legend(loc='upper right')
plt.title('Eğitim ve Doğrulama Kaybı')
plt.xlabel('Epok')
plt.ylabel('Kayıp (Loss)')
plt.show()

# --- 5. Grad-CAM Fonksiyonları ve Görselleştirmesi ---
def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    grad_model = tf.keras.models.Model(
        model.inputs,
        [model.get_layer(last_conv_layer_name).output, model.output]
    )
    
    with tf.GradientTape() as tape:
        last_conv_layer_output, preds = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]
        
    grads = tape.gradient(class_channel, last_conv_layer_output)
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    return heatmap.numpy()

def save_and_display_gradcam(img_path, heatmap, alpha=0.4):
    img = tf.keras.preprocessing.image.load_img(img_path, target_size=IMG_SIZE)
    img = tf.keras.preprocessing.image.img_to_array(img)
    
    heatmap = np.uint8(255 * heatmap)
    jet = cm.get_cmap("jet")
    jet_colors = jet(np.arange(256))[:, :3]
    jet_heatmap = jet_colors[heatmap]
    jet_heatmap = tf.keras.preprocessing.image.array_to_img(jet_heatmap)
    jet_heatmap = jet_heatmap.resize((img.shape[1], img.shape[0]))
    jet_heatmap = tf.keras.image.img_to_array(jet_heatmap)
    
    superimposed_img = jet_heatmap * alpha + img
    superimposed_img = tf.keras.image.array_to_img(superimposed_img)
    
    plt.imshow(superimposed_img)
    plt.axis("off")
    plt.show()

print("\n✅ Grad-CAM görselleştirmesi başlıyor...")
try:
    # En son Evrişim katmanının adını dinamik olarak bulalım.
    last_conv_layer_name = None
    for layer in reversed(current_model.layers):
        if isinstance(layer, Conv2D):
            last_conv_layer_name = layer.name
            break
    
    if last_conv_layer_name is None:
        raise ValueError("Modelde Conv2D katmanı bulunamadı.")
    
    test_classes = os.listdir(testing_path)
    random_class = np.random.choice(test_classes)
    class_path = os.path.join(testing_path, random_class)
    random_image_name = np.random.choice(os.listdir(class_path))
    random_image_path = os.path.join(class_path, random_image_name)

    img = tf.keras.preprocessing.image.load_img(random_image_path, target_size=IMG_SIZE)
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    
    heatmap = make_gradcam_heatmap(img_array, current_model, last_conv_layer_name)
    
    preprocessed_img_array = img_array / 255.0
    predicted_class_index = np.argmax(current_model.predict(preprocessed_img_array))
    predicted_class_name = class_names[predicted_class_index]

    plt.figure(figsize=(8, 8))
    save_and_display_gradcam(random_image_path, heatmap)
    plt.title(f"Grad-CAM - Tahmin Edilen Sınıf: {predicted_class_name}")
    plt.show()
    print("✅ Grad-CAM görselleştirmesi tamamlandı.")

except Exception as e:
    print(f"Hata: Grad-CAM görselleştirmesi sırasında bir sorun oluştu.")
    print(f"Hata Detayı: {e}")
```

### **Veri Ön İşleme (Data Preprocessing)**

Bu aşama, ham verinin modelin anlayabileceği ve işleyebileceği bir formata dönüştürülmesini içerir. Veri setimizi iki ana klasörden (`Training` ve `Testing`) yüklemek için **`tf.keras.utils.image_dataset_from_directory`** fonksiyonunu kullanıyoruz. Bu fonksiyon, veriyi otomatik olarak etiketleyerek ve `tf.data.Dataset` formatına dönüştürerek derin öğrenme modelleri için ideal bir yapı sunar.

**Veri Augmentasyonu (Data Augmentation)**, özellikle küçük veri setlerinde aşırı öğrenmeyi (overfitting) azaltmak için kritik bir tekniktir. `preprocess_train` fonksiyonunda, her eğitim görüntüsüne rastgele **yatay çevirme**, **parlaklık** ve **kontrast** değişiklikleri uygulayarak modelin daha sağlam ve genellenebilir olmasını sağlıyoruz. Görüntü piksellerini 0-255 aralığından 0-1 aralığına **normalize etmek**, modelin daha hızlı ve kararlı bir şekilde yakınsamasına yardımcı olur.

---

### **CNN Modelinin Oluşturulması ve Eğitilmesi**

Modelimiz, beyin tümörlerini sınıflandırmak için bir **Evrişimsel Sinir Ağı (CNN)** mimarisi kullanır. Bu mimari, görüntülerden otomatik olarak özellik çıkarmada son derece etkilidir.

Modelin ana bileşenleri şunlardır:
- **`Conv2D` (Evrişim Katmanları)**: Görüntüdeki kenar, doku ve desen gibi hiyerarşik özellikleri öğrenir. Farklı katmanlar, farklı seviyelerde soyut özellikler çıkarır.
- **`MaxPooling2D` (Havuzlama Katmanları)**: Özellik haritalarının boyutunu küçülterek hesaplama yükünü azaltır ve modelin konum değişikliklerine karşı daha dayanıklı olmasını sağlar.
- **`Dropout`**: Aşırı öğrenmeyi önlemek için eğitim sırasında nöronların rastgele bir kısmını "dışarıda bırakır".
- **`Dense` (Tam Bağlantılı Katmanlar)**: Evrişim katmanlarından çıkarılan özellikler üzerinde son sınıflandırma işlemini gerçekleştirir.
- **`Optimizer`**: **`Adam`** optimizer'ı, eğitim sürecini optimize ederek modelin en iyi sonuçlara ulaşmasını sağlar.
- **`Callbacks` (Geri Çağrımlar)**: Eğitim sürecini otomatik olarak izleyen ve kontrol eden araçlardır. **`EarlyStopping`** ile modelin gereksiz yere eğitime devam etmesini engellerken, **`ModelCheckpoint`** en iyi performans gösteren modeli kaydeder.

---

### **Model Değerlendirmesi ve Performans Analizi**

Model eğitildikten sonra, daha önce görmediği test verileri üzerinde ne kadar iyi performans gösterdiğini değerlendiririz. Bu bölümde kullanılan ana araçlar şunlardır:

- **`Loss` ve `Accuracy` Grafikleri**: Modelin eğitim ve doğrulama seti üzerindeki kayıp ve başarımlarının epok bazında değişimini görselleştirir. Bu grafikler, modelin aşırı veya eksik öğrenme durumunu anlamak için çok önemlidir.
- **`Confusion Matrix` (Karmaşıklık Matrisi)**: Modelin yaptığı tahminleri, gerçek sınıflarla karşılaştıran bir tablo sunar. Hangi sınıfların doğru tahmin edildiğini ve hangilerinin birbiriyle karıştırıldığını net bir şekilde gösterir.
- **`Classification Report` (Sınıflandırma Raporu)**: Her bir sınıf için **kesinlik (precision)**, **geri çağırma (recall)** ve **F1-skoru** gibi detaylı metrikleri sağlar. Bu, modelin her bir sınıf için ne kadar iyi performans gösterdiğini anlamamıza yardımcı olur.

---

### **Grad-CAM Görselleştirmesi**

Modelimiz "kara kutu" bir yapıya sahip olabilir, ancak **Grad-CAM (Gradient-weighted Class Activation Mapping)** tekniği sayesinde bu kara kutunun içine bakabiliriz.

Bu teknik, modelin bir tahmin yaparken görüntünün hangi bölgelerine odaklandığını görselleştirir. Grad-CAM, tahmin edilen sınıfın gradyanlarını son evrişim katmanının çıktılarıyla birleştirerek bir **ısı haritası** oluşturur. Bu harita, modelin kararında en etkili olan pikselleri yüksek yoğunluklu renklerle (genellikle kırmızı) gösterir. 

Bu görselleştirme, modelimizin gerçekten tümör bölgesine mi yoksa arka plan gürültüsü gibi alakasız yerlere mi baktığını anlamamıza olanak tanır. Böylece, modelin güvenilirliğini ve mantıklı kararlar alıp almadığını doğrulayabiliriz.
